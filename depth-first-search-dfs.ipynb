```python
class Tree:
    def __init__(self, value):
        self.value = value
        self.right = None
        self.left = None
```

Problem: Maximum Depth of a Binary Tree

Given the binary tree:

        1
       / \
      2   3
     / \
    4   5

Return the maximum depth of the tree.

Expected output: 3

```python
def findMaximumDepth(root):
    if root is None:
        return 0
    left = findMaximumDepth(root.left)
    right = findMaximumDepth(root.right)
    return 1 + max(left, right)

root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)

print(findMaximumDepth(root))
```

Problem: Minimum Depth of a Binary Tree

Given this tree:

        1
       / \
      2   3
     /     \
    4       5
             \
              6

Return the minimum number of nodes from the root to the nearest leaf.

For this tree:

1 → 2 → 4 = 3             

```python
def findMinimumDepth(root):
    if root is None:
        return 0
    left = findMinimumDepth(root.left)
    right = findMinimumDepth(root.right)
    if left == 0:
        return 1 + right
    if right == 0:
        return 1 + left
    return 1 + min(left, right)
```

Problem: Sum of All Nodes

Given this binary tree:

        1
       / \
      2   3
     / \
    4   5

Return the sum of all node values.

Output = 15

```python
def findSum(root):
    if root is None:
        return 0
    # print(root.value)
    left = findSum(root.left)
    # print('left : ', left)
    right = findSum(root.right)
    return root.value + left + right

root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)
print(findSum(root))
```

Problem: Count Leaf Nodes

Given this binary tree:

        1
       / \
      2   3
     / \   \
    4   5   6

eturns the number of leaf nodes.

A leaf node is a node that has no left and no right child

Here the leaf nodes are: 4, 5, 6

So the expected output is: 3

```python
def countLeafNodes(root):
    if root is None:
        return 0
    if root.left is None and root.right is None:
        return 1
    left = countLeafNodes(root.left)
    right = countLeafNodes(root.right)
    
    return left + right

root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)
print(countLeafNodes(root))    
```

Problem: Find the Maximum Value in a Binary Tree

Given:

        10
       /  \
      5    20
     / \     \
    3   8     15

Return the largest value in the tree.

Expected output: 20

```python
def findMaximumNode(root):
    if root is None:
        return 0
    left = findMaximumNode(root.left)
    right = findMaximumNode(root.right)
    
    return max(root.value, left, right)

root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)
print(findMaximumNode(root)) 
```

Problem 6 — Find a Node in a Binary Tree

Given:

        1
       / \
      2   3
     / \
    4   5

Return True if the target exists anywhere in the tree, otherwise False.

```python
def findNode(root, target):
    if root is None:
        return False
    if root.value == target:
        return True
    left = findNode(root.left, target)
    right = findNode(root.right, target)
    return left or right
root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)
print(findNode(root, 3)) 
```
