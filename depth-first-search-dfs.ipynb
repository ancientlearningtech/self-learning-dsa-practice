Problem: Maximum Depth of a Binary Tree

Given the binary tree:

        1
       / \
      2   3
     / \
    4   5

Return the maximum depth of the tree.

Expected output: 3

```python
def findMaximumDepth(root):
    if root is None:
        return 0
    left = findMaximumDepth(root.left)
    right = findMaximumDepth(root.right)
    return 1 + max(left, right)

root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)

print(findMaximumDepth(root))
```

Problem: Minimum Depth of a Binary Tree

Given this tree:

        1
       / \
      2   3
     /     \
    4       5
             \
              6

Return the minimum number of nodes from the root to the nearest leaf.

For this tree:

1 → 2 → 4 = 3             

```python
def findMinimumDepth(root):
    if root is None:
        return 0
    left = findMinimumDepth(root.left)
    right = findMinimumDepth(root.right)
    if left == 0:
        return 1 + right
    if right == 0:
        return 1 + left
    return 1 + min(left, right)
```

Problem: Sum of All Nodes

Given this binary tree:

        1
       / \
      2   3
     / \

Return the sum of all node values.

Problem 6 — Find a Node in a Binary Tree

Given:

        1
       / \
      2   3
     / \
    4   5

Return True if the target exists anywhere in the tree, otherwise False.

```python
def findNode(root, target):
    if root is None:
        return False
    if root.value == target:
        return True
    left = findNode(root.left, target)
    right = findNode(root.right, target)
    return left or right
root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)
print(findNode(root, 3)) 
```

Problem: Find a Root-to-Leaf Path with a Given Sum

Given this tree:

        5
       / \
      4   8
     /   / \
    3   6   9

Return `True` if there is a root-to-leaf path whose values add up to targetSum; otherwise return `False`.

```python
def hasPathSum(root, targetSum):
    if root is None:
        return False
    if root.left is None and root.right is None:
        return root.value == targetSum
    left = hasPathSum(root.left, targetSum - root.value)
    right = hasPathSum(root.right, targetSum - root.value)
    return left or right

root = Tree(5)
root.left = Tree(4)
root.right = Tree(8)
root.left.left = Tree(3)
root.right.left = Tree(6)
root.right.right = Tree(9)
print(hasPathSum(root, 12))
```

Problem 9 — Find a Root-to-Leaf Path

Given:

        5
       / \
      4   8
     /   / \
    3   6   9

Return the root-to-leaf path containing target.

For example:

target = 6

Expected: [5, 8, 6]

```python
def findPath(root, target):
    if root is None:
        return []
    if root.left is None and root.right is None:
        if root.value == target:
            return [root.value]
        return []
    left = findPath(root.left, target)
    print('left: ', left)
    if left:
        return [root.value] + left
    right = findPath(root.right, target)
    print('right: ', right)
    if right:
        return [root.value] + right
    return []

root = Tree(5)
root.left = Tree(4)
root.right = Tree(8)
root.left.left = Tree(3)
root.right.left = Tree(6)
root.right.right = Tree(9)
print(findPath(root, 6))
```

Problem: Find the Maximum Difference Between a Node and Its Descendant

Given:

        5
       / \
      4   8
     /   / \
    3   6   9

We want the largest difference between a node and any value in its subtree.

For example:

8 → 6   difference = 2

8 → 9   difference = 1

5 → 3   difference = 2

So for this tree:

output → 2

```python
def findMaximumDifference(root):

    def dfs(node):
        if node is None:
            return (float('inf'), float('-inf'), 0)

        if node.left is None and node.right is None:
            return (node.value, node.value, 0)

        left = dfs(node.left)
        right = dfs(node.right)

        leftMinimum = left[0]
        leftMaximum = left[1]
        leftMaximumDifference = left[2]

        rightMinimum = right[0]
        rightMaximum = right[1]
        rightMaximumDifference = right[2]

        minimum = min(node.value, leftMinimum, rightMinimum)
        maximum = max(node.value, leftMaximum, rightMaximum)

        currentDifference = 0

        if node.left is not None:
            currentDifference = max(
                currentDifference,
                abs(node.value - leftMinimum),
                abs(node.value - leftMaximum)
            )

        if node.right is not None:
            currentDifference = max(
                currentDifference,
                abs(node.value - rightMinimum),
                abs(node.value - rightMaximum)
            )

        maximumDifference = max(
            currentDifference,
            leftMaximumDifference,
            rightMaximumDifference
        )

        return (minimum, maximum, maximumDifference)

    return dfs(root)[2]


root = Tree(5)
root.left = Tree(4)
root.right = Tree(8)
root.left.left = Tree(3)
root.right.left = Tree(6)
root.right.right = Tree(9)

print(findMaximumDifference(root))
```

Problem: Search for a Value in a Binary Tree

Given the root of a binary tree and a target value, return True if the target exists in the tree. Otherwise, return False.

Example:

        5
       / \
      4   8
     /   / \
    3   6   9

Target: 6

Output: True

```python
def containsValue(root, target):
    if root is None:
        return False
    if root.value == target:
        return True
    print(root.value)
    left = containsValue(root.left, target)
    if left:
        return True
    right = containsValue(root.right, target)
    return right

root = Tree(5)
root.left = Tree(4)
root.right = Tree(8)
root.left.left = Tree(3)
root.right.left = Tree(6)
root.right.right = Tree(9)

print(containsValue(root, 6))
print(containsValue(root, 7))
print(containsValue(root, 3))
```
