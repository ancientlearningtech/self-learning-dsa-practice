```python
class Tree:
    def __init__(self, value):
        self.value = value
        self.right = None
        self.left = None
```

Problem: Maximum Depth of a Binary Tree

Given the binary tree:

        1
       / \
      2   3
     / \
    4   5

Return the maximum depth of the tree.

Expected output: 3

```python
def findMaximumDepth(root):
    if root is None:
        return 0
    left = findMaximumDepth(root.left)
    right = findMaximumDepth(root.right)
    return 1 + max(left, right)

root = Tree(1)
root.left = Tree(2)
root.right = Tree(3)
root.left.left = Tree(4)
root.left.right = Tree(5)

print(findMaximumDepth(root))
```
