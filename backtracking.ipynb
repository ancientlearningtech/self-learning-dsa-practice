Backtracking: Letter Case Permutation

Given a string containing letters and digits, generate all possible strings by changing the case of the letters.

Example

input = "a1b2"

Possible output:

    [
        "a1b2",
        "a1B2",
        "A1b2",
        "A1B2"
    ]

```python
def generateLetterCasePermutations(text, index, current, result):
    if len(current) == len(text):
        return result.append(current)
    if text[index].isalpha():
        current += text[index].lower()
        generateLetterCasePermutations(text, index + 1, current, result)
        current = current[:-1]
        
        current += text[index].upper()
        generateLetterCasePermutations(text, index + 1, current, result)
        current = current[:-1]
    else:
        current += text[index]
        generateLetterCasePermutations(text, index + 1, current, result)
        current = current[:-1]
    return result

print(generateLetterCasePermutations('a1b2', 0, '', []))
```

problem: Generate Binary Strings

Given n, generate all binary strings of length n.

For:

n = 2

Output:

    ["00", "01", "10", "11"]

```python
def generateBinaryStrings(n, current, result):
    if len(current) == n:
        return result.append(current)
    
    current += "0"
    generateBinaryStrings(n, current, result)
    current = current[:-1]
    
    current += "1"
    generateBinaryStrings(n, current, result)
    current = current[:-1]
    
    return result

print(generateBinaryStrings(2, "", []))
```

Problem: Generate All Subsets

Given an array of distinct numbers, generate all possible subsets.

Example

numbers = [1, 2]

Output:

    [
        [],
        [1],
        [2],
        [1, 2]
    ]

```python
def generateSubsets(numbers, index, current, result):
    if len(numbers) == index:
        return result.append(current)
    generateSubsets(numbers, index + 1, current + [numbers[index]], result)
    generateSubsets(numbers, index + 1, current, result)
    return result

print(generateSubsets([1,2], 0, [], []))
```

Problem: Generate All Binary Strings Without Consecutive 1s

Given n, generate all binary strings of length n that do not contain consecutive 1s.

For:

n = 3

Expected:

[
    "000",
    "001",
    "010",
    "100",
    "101"
]

```python
def generateValidBinaryStrings(n, current, result):
    if len(current) == n:
        return result.append(current)
    current += '0'
    generateValidBinaryStrings(n, current, result)
    current = current[:-1]
    if current == '' or current[-1] != '1':
        current += '1'
        generateValidBinaryStrings(n, current, result)
        current = current[:-1]
    return result
    
print(generateValidBinaryStrings(3, '', []))
print(generateValidBinaryStrings(2, '', []))
```

n = 3

Valid output:

    "BAB",
    "BBA",
    "BBB"

"AAA"

Problem: Generate All Permutations of "ABC"

Given a string, generate all possible arrangements of its characters.

For:

text = "ABC"

Expected:

[
    "ABC",
    "ACB",
    "BAC",
    "BCA",
    "CAB",
    "CBA"
]

```python
def generateStringPermutations(text, current, result):
    if len(text) == 0:
        result.append(current)
        return result
    for i in range(len(text)):
        current += text[i]
        generateStringPermutations(text[:i] + text[i + 1:], current, result)
        current = current[:-1]
    return result
    
print(generateStringPermutations("ABC", "", []))
```

Problem: Generate All Strings of `A` and `B` With Exactly `k` A's

Given:

n = 3

k = 2

Generate all strings of length 3 containing exactly two As.

Expected:

["AAB", "ABA", "BAA"]

```python
def generateStringsWithKAs(n, k, current, aCount, result):
    if len(current) == n:
        if aCount == k:
            result.append(current)
        return result
    if aCount < k:
        current += 'A'
        aCount += 1
        generateStringsWithKAs(n, k, current, aCount, result)
        current = current[:-1]
        aCount -= 1
    
    current += 'B'
    generateStringsWithKAs(n, k, current, aCount, result)
    current = current[:-1]
    return result

print(generateStringsWithKAs(3, 2, "", 0, []))
```
