Problem 1 – Merge Intervals
Problem Statement

You are given an array of intervals where each interval is represented as:

[start, end]

Each interval represents all numbers from start to end (inclusive).

Your task is to merge all overlapping intervals and return a new list of non-overlapping intervals that covers the same ranges.

Function Signature
def mergeIntervals(intervals):
Input

A list of intervals:

intervals = [[1,3],[2,6],[8,10],[15,18]]
Output

Return the merged list of intervals.

[[1,6],[8,10],[15,18]]

```python
def mergeIntervals(intervals):
    if len(intervals) <= 0:
        return intervals
    intervals.sort()
    mergedIntervals = []    
    
    current = 0
    next = current + 1
    currentEnd = intervals[current][1]
    
    while next < len(intervals):
        # print(current, next, currentEnd)
        if currentEnd >= intervals[next][0]:
            currentEnd = max(currentEnd, intervals[next][1])
            next += 1
        else:
            mergedIntervals.append([intervals[current][0], currentEnd])
            current = next
            next += 1
            currentEnd = intervals[current][1]
    # if next == len(intervals):
    mergedIntervals.append([intervals[current][0], currentEnd])
    
    return mergedIntervals

print(mergeIntervals([[1, 3], [2, 6], [8, 10], [15, 18]]))
# print(mergeIntervals([[1,4],[2,5],[3,6]]))
# print(mergeIntervals([[1,2],[4,5],[7,8]]))
# print(mergeIntervals([[1,10],[3,4],[5,6]]))
# print(mergeIntervals([[1,5]]))
# print(mergeIntervals([]))
# print(mergeIntervals([[1,4],[4,5]]))
# print(mergeIntervals([[1,4],[0,2],[3,5]]))
# print(mergeIntervals([[2,3],[4,5],[6,7],[8,9],[1,10]]))
```

```python
# chatgpt solution
def mergeIntervals(intervals):
    if len(intervals) <= 1:
        return intervals

    intervals.sort(key=lambda interval: interval[0])

    mergedIntervals = []

    currentInterval = intervals[0]

    for nextInterval in intervals[1:]:

        if nextInterval[0] <= currentInterval[1]:
            currentInterval[1] = max(currentInterval[1], nextInterval[1])
        else:
            mergedIntervals.append(currentInterval)
            currentInterval = nextInterval

    mergedIntervals.append(currentInterval)

    return mergedIntervals
```

Problem – Insert Interval

You are given:

A list of non-overlapping intervals sorted by their start values.
A new interval.

Insert the new interval into the list such that:

The intervals remain sorted.
Any overlapping intervals are merged.

Return the updated list of intervals.

Function Signature
def insertInterval(intervals, newInterval):
Example 1
Input
intervals = [[1,3],[6,9]]
newInterval = [2,5]
Output
[[1,5],[6,9]]
Explanation

The new interval overlaps with [1,3], so they merge into [1,5].

```python
def insertInterval(intervals, newInterval):
    mergedIntervals = []
    
    index = 0
    
    while index < len(intervals) and intervals[index][1] < newInterval[0]:
        mergedIntervals.append(intervals[index])
        index += 1
        
    while index < len(intervals) and intervals[index][0] <= newInterval[1]:
        newInterval[0] = min(newInterval[0], intervals[index][0])
        newInterval[1] = max(newInterval[1], intervals[index][1])
        index += 1
        
    mergedIntervals.append(newInterval)
    
    while index < len(intervals):
        mergedIntervals.append(intervals[index])
        index += 1
        
    return mergedIntervals
```
