Problem 1: Maximum Sum Subarray of Size K

Given an array of positive integers and a positive integer k, find the maximum sum of any contiguous subarray of size k.

Example
arr = [2, 1, 5, 1, 3, 2]
k = 3
Output
9

````python
def maxSubarraySum(array, k):
    if k > len(array) or k <= 0:
        return 0

    currentSum = sum(array[:k])
    maxSum = currentSum

    for right in range(k, len(array)):
        currentSum = currentSum - array[right - k] + array[right]
        maxSum = max(maxSum, currentSum)

    return maxSum


print(maxSubarraySum([2, 1, 5, 1, 3, 2], 3))
````

Problem 2

Average of Every Subarray of Size K

Given:

array = [1, 3, 2, 6, -1, 4, 1, 8, 2]
k = 5

Return the average of every contiguous subarray of size k.

Example output:

[2.2, 2.8, 2.4, 3.6, 2.8]

````python
def averageOfAllSubarrays(array, k):
    if k > len(array) or k <= 0:
        return 0
    
    averageOfAllSubarraysList = []
    
    currentSum = sum(array[:k])
    averageOfAllSubarraysList.append(currentSum / k)
    
    for right in range(k, len(array)):
        currentSum = currentSum - array[right - k] + array[right]
        averageOfAllSubarraysList.append(currentSum / k)
        
    return averageOfAllSubarraysList

print(averageOfAllSubarrays([1, 3, 2, 6, -1, 4, 1, 8, 2], 5))
````

Vowels are:

a, e, i, o, u
Example 1

3

print(maxVowelsInSubstring('abciiidef', 3))```

````python
def maxVowelsInSubstring(myString, k):
    if k > len(myString) or k <= 0:
        return 0

    vowels = {'a', 'e', 'i', 'o', 'u'}

    currentVowelCount = 0

    for char in myString[:k]:
        if char in vowels:
            currentVowelCount += 1

    maxVowelCount = currentVowelCount

    for right in range(k, len(myString)):

        if myString[right - k] in vowels:
            currentVowelCount -= 1

        if myString[right] in vowels:
            currentVowelCount += 1

        maxVowelCount = max(maxVowelCount, currentVowelCount)

    return maxVowelCount


print(maxVowelsInSubstring('abciiidef', 3))
````

Problem 4: Maximum Number of Occurrences of '1' in a Window of Size K

Given a binary array (containing only 0 and 1) and an integer k, find the maximum number of 1s present in any contiguous subarray of size k.

Example
array = [1, 0, 1, 1, 0, 1, 0, 1]
k = 3
Output
2

````python
def maxOccurenceOfOne(array, k):
    if k > len(array) or k <= 0:
        return 0
    
    countOnes = 0
    
    for i in range(len(array[:k])):
        if array[i] == 1:
            countOnes += 1
            
    maxCountOfOnes = countOnes
    
    for right in range(k, len(array)):
        if array[right - k] == 1:
            countOnes -= 1
            
        if array[right] == 1:
            countOnes += 1
            
        maxCountOfOnes = max(maxCountOfOnes, countOnes)
        
    return maxCountOfOnes

print(maxOccurenceOfOne([1, 0, 1, 1, 0, 1, 0, 1], 3))
````

Problem: Smallest Subarray With Sum ≥ Target

Given an array of positive integers nums and a positive integer target, return the length of the smallest contiguous subarray whose sum is greater than or equal to target.

If no such subarray exists, return 0.

Example
target = 7
nums = [2, 3, 1, 2, 4, 3]
Output
2

````python
def minSubArrayLength(array, target):
    left = 0
    currentSum = 0
    minLength = float('inf')
    
    for right in range(len(array)):
        currentSum += array[right]
         
        while currentSum >= target:
            minLength = min(minLength, right - left + 1)
            currentSum -= array[left]
            left += 1
            
    if minLength == float('inf'):
        return 0
    
    return minLength

print(minSubArrayLength([2, 3, 1, 2, 4, 3], 7))
````

Problem: Longest Subarray with Sum ≤ K

Given an array of positive integers nums and an integer k, find the length of the longest contiguous subarray whose sum is less than or equal to k.

If no such subarray exists, return 0.

Example
nums = [1, 2, 1, 0, 1, 1, 0]
k = 4
Output
5

Explanation

One of the longest valid subarrays is:

[1, 0, 1, 1, 0]

````python
def maxSubArrayLength(array, k):
    left = 0
    currentSum = 0
    maxLength = 0
    
    for right in range(len(array)):
        currentSum += array[right]
        
        while currentSum > k:
            currentSum -= array[left]
            left += 1
        
        maxLength = max(maxLength, right - left + 1)
            
    return maxLength

print(maxSubArrayLength([1, 2, 1, 0, 1, 1, 0], 4))
````
