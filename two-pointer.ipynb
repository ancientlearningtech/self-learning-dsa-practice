Two Sum II (Sorted Array)

Given a sorted array of integers and a target, return the indices of the two numbers whose sum equals the target.

Example:

numbers = [2, 7, 11, 15]
target = 9

Output:
[0, 1]

Another example:

numbers = [1, 3, 4, 6, 8, 10]
target = 14

Output:
[2, 5]

```python
def findTarget(numbers, target):
    if len(numbers) < 2:
        return []
    
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        currentSum = numbers[left] + numbers[right]
        
        if currentSum > target:
            right -= 1
        elif currentSum < target:
            left += 1
        else:
            return [left, right]
    
    return []

print(findTarget([2,7,11,15], 9))
# print(findTarget([1,3,4,6,8,10], 14))# [0,1]
```

Valid Palindrome

Given a string s, return True if it is a palindrome, otherwise return False.

A palindrome reads the same forwards and backwards.

Example 1
Input:
s = "madam"

Output:
True

```python
def isPalindrome(s):
    left = 0
    right = len(s) - 1
    
    while left < right:
        if s[left] == s[right]:
            left += 1
            right -= 1
        else:
            return False
    
    return True

print(isPalindrome('madam')) # True
# print(isPalindrome('racecar')) # True
# print(isPalindrome('python')) # False
# print(isPalindrome('abba')) # True
```

Remove Duplicates from Sorted Array

Given a sorted array, remove the duplicates in-place such that each unique element appears only once.

Return the number of unique elements.

For our practice, don't worry about modifying the array exactly as LeetCode requires. Our focus is on understanding the pointer movement.

Example 1
Input:
numbers = [1, 1, 2]

Output:
2

The unique elements are:

[1, 2]

```python
def removeDuplicates(numbers):
    left = 0
    right = 1
    
    while right < len(numbers):
        if numbers[right] != numbers[left]:
            numbers[left + 1] = numbers[right]
            left += 1
        right += 1
    
    return numbers[: left + 1]

# print(removeDuplicates([1,1,2])) # [1,2]
# print(removeDuplicates([1,1,2,2,3])) # [1,2,3]
print(removeDuplicates([0,0,1,1,1,2,2,3,3,4])) # [0,1,2,3,4]
```

Move Zeroes

Given an integer array numbers, move all the 0s to the end of the array while maintaining the relative order of the non-zero elements.

Do this in-place.

Example 1
Input:
numbers = [0, 1, 0, 3, 12]

Output:
[1, 3, 12, 0, 0]

```python
def moveZeroesToEnd(numbers):
    left = 0
    right = 0
    
    while right < len(numbers):
        if numbers[right] != 0:
            numbers[left], numbers[right] = numbers[right], numbers[left]
            left += 1
        right += 1
    
    return numbers

# print(moveZeroesToEnd([0,1,0,3,12]))
# print(moveZeroesToEnd([0,0,1]))
# print(moveZeroesToEnd([1, 2, 3]))
print(moveZeroesToEnd([1, 0, 2]))
```

Return the remaining elements (for our practice).

Squares of a Sorted Array
Problem Statement

Given a sorted array of integers (which may include negative numbers), return a new array containing the squares of each number, also sorted in non-decreasing order.

Example 1
Input:
numbers = [-4, -1, 0, 3, 10]

Output:
[0, 1, 9, 16, 100]
Example 2
Input:
numbers = [-7, -3, 2, 3, 11]

Output:
[4, 9, 9, 49, 121]

```python
def getSortedSquares(numbers):
    left = 0
    right = len(numbers) - 1
    squaredNumbers = [0] * len(numbers)
    write = len(numbers) - 1
    
    while left <= right:
        if abs(numbers[left]) > abs(numbers[right]):
            squaredNumbers[write] = pow(numbers[left], 2)
            left += 1
        else:
            squaredNumbers[write] = pow(numbers[right], 2)
            right -= 1
        write -= 1
    
    return squaredNumbers

print(getSortedSquares([-4,-1,0,3,10]))
        
```

Problem

You are given an array heights, where each element represents the height of a vertical line.

Choose two lines such that together with the x-axis they form a container that can hold the maximum amount of water.

Return the maximum amount of water.

Example
heights = [1,8,6,2,5,4,8,3,7]

Output:
49

The maximum water is formed by the lines with heights 8 and 7.

```python
def getMaximumWaterArea(heights):
    left = 0
    right = len(heights) - 1
    maximumContainerwater = 0
    
    while left < right:
        currentMaximumWater = min(heights[left], heights[right])
        calculateMaximumWater = currentMaximumWater * (right - left)
        if heights[left] < heights[right]:
            left += 1
        else:
            right -= 1
        maximumContainerwater = max(maximumContainerwater, calculateMaximumWater) 
    return maximumContainerwater

print(getMaximumWaterArea([1,8,6,2,5,4,8,3,7]))          
```

Problem 8

Given an integer array numbers, return all unique triplets such that:

a + b + c = 0

No duplicate triplets should appear in the answer.

Example
numbers = [-1, 0, 1, 2, -1, -4]

Expected output:

[
    [-1, -1, 2],
    [-1, 0, 1]
]

Order of the triplets doesn't matter.

```python
def getThreeSum(numbers):
    numbers.sort()
    result = []
        
    for fixed in range(len(numbers) - 2):
        if fixed > 0 and numbers[fixed] == numbers[fixed - 1]:
            continue
        
        left = fixed + 1
        right = len(numbers) - 1
        target = -numbers[fixed]

        while left < right:
            currentSum = numbers[left] + numbers[right]
            if currentSum < target:
                left += 1
            elif currentSum > target:
                right -= 1
            else:
                result.append([numbers[fixed], numbers[left], numbers[right]])
                
                left += 1
                right -= 1
                while left < right and numbers[left] == numbers[left - 1]:
                    left += 1
                while left < right and numbers[right] == numbers[right + 1]:
                    right -= 1
    
    return result
     
print(getThreeSum([-1, 0, 1, 2, -1, -4]))
```

Problem 9

Given an integer array numbers and an integer target, return the sum of three integers whose sum is closest to the target.

You may assume that exactly one solution exists.

Example
numbers = [-1, 2, 1, -4]
target = 1

Output:

2

```python
def getClosestThreeSum(numbers, target):
    numbers.sort()
    # print(numbers)
    
    closestSum = numbers[0] + numbers[1] + numbers[2]
    
    for fixed in range(len(numbers) - 2):
        if fixed > 0 and numbers[fixed] == numbers[fixed - 1]:
            continue
        left = fixed + 1
        right = len(numbers) - 1
        
        while left < right:
            currentSum = numbers[fixed] + numbers[left] + numbers[right]
            if abs(currentSum - target) < abs(closestSum - target):
                closestSum = currentSum
            if currentSum < target:
                left += 1
            elif currentSum > target:
                right -= 1
            else:
                return currentSum
                
    return closestSum

print(getClosestThreeSum([-1,2,1,-4], 1))
```

Problem 10

Given an array of integers numbers and an integer target, return all unique quadruplets such that:

a + b + c + d = target

No duplicate quadruplets should be returned.

Example
numbers = [1, 0, -1, 0, -2, 2]
target = 0

Expected output:

[
    [-2, -1, 1, 2],
    [-2,  0, 0, 2],
    [-1,  0, 0, 1]
]

```python
def getFourSum(numbers, target):
    numbers.sort()
    result = []
    
    for first in range(len(numbers) - 3):
        if first > 0 and numbers[first] == numbers[first - 1]:
            continue
        for second in range(first + 1, len(numbers) - 2):
            if second > first + 1 and numbers[second] == numbers[second - 1]:
                continue
            left = second + 1
            right = len(numbers) - 1
            
            while left < right:
                currentSum = numbers[first] + numbers[second] + numbers[left] + numbers[right]
                if currentSum < target:
                    left += 1
                elif currentSum > target:
                    right -= 1
                else:
                    result.append([numbers[first], numbers[second], numbers[left], numbers[right]])
                
                    left += 1
                    right -= 1
                    
                    while left < right and numbers[left] == numbers[left - 1]:
                        left += 1
                    while left < right and numbers[right] == numbers[right + 1]:
                        right -= 1
    
    return result

print(getFourSum([1,0,-1,0,-2,2], 0))            
```

Problem 11: 

Trapping Rain Water (Hard)

Problem Statement

You are given an array height where each element represents the height of a bar.

After it rains, water gets trapped between the bars.

Return the total amount of water trapped.

Example 1
Input:
height = [0,1,0,2,1,0,1,3,2,1,2,1]

Output:
6

```python
def getTrappedRainWater(heights):
    left = 0
    right = len(heights) - 1

    leftMax = 0
    rightMax = 0

    totalWater = 0

    while left < right:
        if heights[left] < heights[right]:
            leftMax = max(leftMax, heights[left])
            totalWater += leftMax - heights[left]
            left += 1
        else:
            rightMax = max(rightMax, heights[right])
            totalWater += rightMax - heights[right]
            right -= 1
            
    return totalWater

print(getTrappedRainWater([0,1,0,2,1,0,1,3,2,1,2,1]))
# print(getTrappedRainWater([3, 0, 2, 0, 4]))
# print(getTrappedRainWater([3, 0, 5, 0, 4]))
# print(getTrappedRainWater([3, 2, 5, 2, 4]))
```

Problem

You are given an array people where:

people[i] = weight of the ith person

You are also given:

limit

Each boat can carry at most two people, and the combined weight of the people in a boat cannot exceed limit.

Return the minimum number of boats required to rescue everyone.

Example 1
people = [1,2]
limit = 3

Output

1

Explanation

Boat 1 → (1,2)

```python
def maximumBoatsRequired(people, limit):
    people.sort()
    
    left = 0
    right = len(people) - 1
    boats = 0
    
    while left < right:
        if people[left] == limit:
            boats += 1
            left += 1
        if people[right] == limit:
            boats += 1
            right -= 1
        if people[left] + people[right] <= limit:
            boats += 1
            left += 1
            right -= 1
        elif people[left] < limit:
            boats += 1
            left += 1
        elif people[right] < limit:
            boats += 1
            right -= 1
        if left == right:
            boats += 1
    
    return boats

print(maximumBoatsRequired([3,2,2,1], 3))
print(maximumBoatsRequired([1,2], 3))
print(maximumBoatsRequired([1,2,2], 3))
print(maximumBoatsRequired([1,1,2,2], 3))
print(maximumBoatsRequired([2,2,2,2], 3))
```

Problem: Sort Colors (Dutch National Flag)

You are given an array containing only the numbers:

0, 1, 2

where:

0 represents Red
1 represents White
2 represents Blue

Your task is to sort the array in-place so that all:

0s come first,
then all 1s,
then all 2s.

You cannot use the built-in sort function.

Example 1
numbers = [2, 0, 2, 1, 1, 0]

Output

[0, 0, 1, 1, 2, 2]

```python
def sortColors(numbers):
    low = 0
    mid = 0
    high = len(numbers) - 1
    
    while mid <= high:
        if numbers[mid] == 0:
            numbers[low], numbers[mid] = numbers[mid], numbers[low]
            low += 1
            mid += 1
        elif numbers[mid] == 1:
            mid += 1
        else:
            numbers[high], numbers[mid] = numbers[mid], numbers[high]
            high -= 1
            
    return numbers

print(sortColors([2,0,2,1,1,0]))        
```

Problem: Linked List Cycle (Easy)
Problem

Given the head of a singly linked list, determine whether the linked list contains a cycle.

A cycle exists if a node in the list can be reached again by continuously following the next pointers.

Return:

True if the linked list contains a cycle.
False otherwise.
Example 1
Input:

1 → 2 → 3 → 4
    ↑       ↓
    ← ← ← ←

Output:

True

Explanation:

The last node points back to the second node, creating a cycle.

Example 2
Input:

1 → 2 → 3 → 4 → None

Output:

False

Explanation:

The last node points to None, so there is no cycle.

```python
class ListNode:
    def __init__(self, value):
        self.value = value
        self.next = None
        
def hasCycle(head):
    slow = head
    fast = head
    
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        
        if slow == fast:
            return True

    return False
```

A happy number is defined using the following process:

Start with any positive integer n.

Return:

True if the number is a happy number.

True

Input:

2

Output:

False

```python
# Helper function to extract sum of squares of the digits

def getSumOfSquares(number):
    if number < 10:
        return pow(number, 2)
    return pow((number % 10), 2) + getSumOfSquares(number // 10)
        
```

```python
def isHappyNumber(number):
    slow = getSumOfSquares(number)
    fast = getSumOfSquares(getSumOfSquares(number))
    
    while slow != fast:
        slow = getSumOfSquares(slow)
        fast = getSumOfSquares(getSumOfSquares(fast))
    
    return slow == 1

print(isHappyNumber(19))
print(isHappyNumber(1))
print(isHappyNumber(2))
print(isHappyNumber(68))
```

Problem: Find the Start of the Cycle (Medium)

This is one of the classic interview questions based on Floyd's Cycle Detection Algorithm.

Problem Statement

You are given the head of a singly linked list.

If the linked list contains a cycle, return the node where the cycle begins.

If there is no cycle, return None.

You are not allowed to modify the linked list.

Example 1
Input:

1 → 2 → 3 → 4 → 5
      ↑       ↓
      ← ← ← ←

Output:

Node with value 2

Explanation:

The last node points back to the node with value 2, so the cycle begins at node 2.

```python
class ListNode:
    def __init__(self, value):
        self.value = value
        self.next = None


def getCycleStartNode(head):
    slowPointer = head
    fastPointer = head

    # Detect whether a cycle exists
    while fastPointer is not None and fastPointer.next is not None:
        slowPointer = slowPointer.next
        fastPointer = fastPointer.next.next

        if slowPointer == fastPointer:
            break
    else:
        return None

    # Find the start of the cycle
    slowPointer = head

    while slowPointer != fastPointer:
        slowPointer = slowPointer.next
        fastPointer = fastPointer.next

    return slowPointer
```

Problem: Circular Array Loop (Medium)

This is based on LeetCode 457 and is considered one of the classic applications of Floyd's algorithm.

Problem:

You are given a circular array of integers called numbers.

Each element tells you how many steps to move.

A positive number means move forward.
A negative number means move backward.
The array is circular, so moving past the last index wraps around to the beginning, and moving before index 0 wraps around to the end.

Determine whether there exists a cycle that satisfies all of these conditions:

The cycle contains more than one element.
Every movement in the cycle is in the same direction:
either all positive, or
all negative.

Return:

True if such a cycle exists.
False otherwise.
Example 1
Input:

numbers = [2, -1, 1, 2, 2]

Output:

True

```python
def getNextIndex(numbers, currentIndex, isForward):
    currentDirection = numbers[currentIndex] >= 0

    # Direction changed
    if currentDirection != isForward:
        return -1

    nextIndex = (currentIndex + numbers[currentIndex]) % len(numbers)

    # One-element cycle is not allowed
    if nextIndex == currentIndex:
        return -1

    return nextIndex


def hasCircularArrayLoop(numbers):
    for startIndex in range(len(numbers)):
        isForward = numbers[startIndex] >= 0

        slowPointer = startIndex
        fastPointer = startIndex

        while True:
            slowPointer = getNextIndex(numbers, slowPointer, isForward)
            if slowPointer == -1:
                break

            fastPointer = getNextIndex(numbers, fastPointer, isForward)
            if fastPointer == -1:
                break

            fastPointer = getNextIndex(numbers, fastPointer, isForward)
            if fastPointer == -1:
                break

            if slowPointer == fastPointer:
                return True

    return False
```

Problem: Find the Length of the Cycle (Easy)
Problem:

You are given the head of a singly linked list.

If the linked list contains a cycle, return the length of the cycle.

If there is no cycle, return 0.

You are not allowed to modify the linked list.

Example 1
1 → 2 → 3 → 4 → 5
      ↑       ↓
      ← ← ← ←

The cycle is:

2 → 3 → 4 → 5 → 2

Length:

4

Output:

4

```python
class ListNode:
    def __init__(self, value):
        self.value = value
        self.next = None


def getCycleLength(head):
    slowPointer = head
    fastPointer = head

    # Step 1: Detect the cycle
    while fastPointer is not None and fastPointer.next is not None:
        slowPointer = slowPointer.next
        fastPointer = fastPointer.next.next

        if slowPointer == fastPointer:
            # Step 2: Count the cycle length
            cycleLength = 1
            currentNode = slowPointer.next

            while currentNode != slowPointer:
                cycleLength += 1
                currentNode = currentNode.next

            return cycleLength

    return 0
```
