Problem Statement

Given an integer array numbers and an integer windowSize, find the maximum sum among all contiguous subarrays of size windowSize.

A contiguous subarray means the elements are next to each other in the array.

Return the maximum sum.

Example 1

Input

numbers = [2, 1, 5, 1, 3, 2]
windowSize = 3

Explanation

Subarrays of size 3:

[2, 1, 5] → 8
[1, 5, 1] → 7
[5, 1, 3] → 9
[1, 3, 2] → 6

Output

9

```python
def findMaxSubarraySum(nums, k):
    if len(nums) < k:
        return 0
    
    left = 0
    totalSum = 0
    
    for i in range(k):
        totalSum += nums[i]
    
    maxSum = totalSum
    
    for right in range(k, len(nums)):
        totalSum = totalSum - nums[left] + nums[right]
        left += 1
        maxSum = max(maxSum, totalSum)
        
    return maxSum

print(findMaxSubarraySum([2,1,5,1,3,2], 3))
```

Problem Statement

Given an integer array numbers and an integer windowSize, find the maximum average of any contiguous subarray of size windowSize.

Return the maximum average.

Example 1

Input

numbers = [1, 12, -5, -6, 50, 3]
windowSize = 4

Windows

[1, 12, -5, -6]  → Sum = 2   → Average = 0.5
[12, -5, -6, 50] → Sum = 51  → Average = 12.75
[-5, -6, 50, 3]  → Sum = 42  → Average = 10.5

Output

12.75

```python
def findMaxAverageSubarray(numbers, windowSize):
    if len(numbers) < windowSize:
        return 0
    
    left = 0
    currentSum = 0
    
    
    for elem in range(windowSize):
        currentSum += numbers[elem]
    maxAverage = currentSum / windowSize
    
    for right in range(windowSize, len(numbers)):
        currentSum = currentSum - numbers[left] + numbers[right]
        left += 1
        maxAverage = max(maxAverage, currentSum / windowSize)
        
    return maxAverage
    
print(findMaxAverageSubarray([1,12,-5,-6,50,3], 4))
print(findMaxAverageSubarray([-5, -4, -3, -2], 2))
```

Problem Statement

Given a string text and an integer windowSize, return the maximum number of vowels present in any substring of length windowSize.

The vowels are:

a, e, i, o, u

Assume the string contains only lowercase English letters.

Example 1

Input

text = "abciiidef"
windowSize = 3

Substrings

"abc" → 1 vowel
"bci" → 1 vowel
"cii" → 2 vowels
"iii" → 3 vowels
"iid" → 2 vowels
"ide" → 2 vowels
"def" → 1 vowel

Output

3

```python
def findMaxVowelsInSubstring(text, windowSize):
    if len(text) < windowSize:
        return 0
    vowels = "aeiou"
    maxVowels = 0
    for char in range(windowSize):
        if text[char] in vowels:
            maxVowels += 1

    left = 0
    currentVowels = maxVowels
    
    for right in range(windowSize, len(text)):
        if text[left] in vowels:
            currentVowels -= 1
        if text[right] in vowels:
            currentVowels += 1
        left += 1
        maxVowels = max(maxVowels, currentVowels)
    
    return maxVowels
        
print(findMaxVowelsInSubstring("abciiidef", 3))
print(findMaxVowelsInSubstring("aeiou", 2))
```

Problem Statement

Given a string text and a string pattern, determine whether any permutation of pattern exists as a contiguous substring in text.

Return:

True if such a substring exists.
False otherwise.
Example 1

Input

text = "oidbcaf"
pattern = "abc"

Explanation

The substring:

"bca"

is a permutation of "abc".

Output

True

```python
def containsPermutation(text, pattern):
    if len(text) < len(pattern):
        return False
    
    sourceDict = dict()
    targetDict = dict()
    
    for char in pattern:
        targetDict[char] = targetDict.get(char, 0) + 1
        
    left = 0
    
    for right in range(len(text)):
        sourceDict[text[right]] = sourceDict.get(text[right], 0) + 1
        
        while right - left + 1 > len(pattern):
            sourceDict[text[left]] -= 1
            if sourceDict[text[left]] == 0:
                del sourceDict[text[left]]
            left += 1
            
        if sourceDict == targetDict:
            return True
    
    return False
    
print(containsPermutation('aaacb', 'abc'))
# print(containsPermutation('abcdef', 'abc'))
# print(containsPermutation('abcdef', 'bcd'))
# print(containsPermutation('abcdef', 'acd'))
# print(containsPermutation('oidbcaf', 'abc'))
```

```python
# Better solution but the previous solution is accepted too

def containsPermutation(text, pattern):
    if len(text) < len(pattern):
        return False

    windowSize = len(pattern)

    windowFrequency = {}
    patternFrequency = {}

    # Build the pattern frequency map
    for character in pattern:
        patternFrequency[character] = patternFrequency.get(character, 0) + 1

    left = 0

    for right in range(len(text)):
        # Add the incoming character
        windowFrequency[text[right]] = windowFrequency.get(text[right], 0) + 1

        # Shrink once if the window becomes too large
        if right - left + 1 > windowSize:
            outgoingCharacter = text[left]

            windowFrequency[outgoingCharacter] -= 1

            if windowFrequency[outgoingCharacter] == 0:
                del windowFrequency[outgoingCharacter]

            left += 1

        # Compare only when the window size is exactly equal to the pattern length
        if right - left + 1 == windowSize:
            if windowFrequency == patternFrequency:
                return True

    return False

print(containsPermutation('aaacb', 'abc'))
# print(containsPermutation('abcdef', 'abc'))
# print(containsPermutation('abcdef', 'bcd'))
# print(containsPermutation('abcdef', 'acd'))
# print(containsPermutation('oidbcaf', 'abc'))
```

Problem Statement

Given a string text and an integer windowSize, find the maximum number of distinct characters present in any substring of length windowSize.

Return the maximum number of distinct characters among all windows.

Example 1

Input

text = "aabcbcdbca"
windowSize = 3

Windows

"aab" → {a, b}           → 2 distinct
"abc" → {a, b, c}        → 3 distinct
"bcb" → {b, c}           → 2 distinct
"cbc" → {b, c}           → 2 distinct
"bcd" → {b, c, d}        → 3 distinct
"cdb" → {c, d, b}        → 3 distinct
"dbc" → {d, b, c}        → 3 distinct
"bca" → {b, c, a}        → 3 distinct

Output

3

```python
def findMaxDistinctCharacters(text, windowSize):
    if len(text) < windowSize:
        return 0
    
    left = 0
    distinctCharacterLength = 0
    targetDict = {}
    
    for char in range(windowSize):
        targetDict[text[char]] = targetDict.get(text[char], 0) + 1
    
    distinctCharacterLength = len(targetDict)
    
    for right in range(windowSize, len(text)):
        targetDict[text[right]] = targetDict.get(text[right], 0) + 1
        targetDict[text[left]] -= 1
        if targetDict[text[left]] == 0:
            del targetDict[text[left]]
        left += 1
        distinctCharacterLength = max(distinctCharacterLength, len(targetDict))
        
    return distinctCharacterLength
    
print(findMaxDistinctCharacters("aabcbcdbca", 3)) # 3
# print(findMaxDistinctCharacters("aabcbcdb", 3)) # 3
# print(findMaxDistinctCharacters("aaaaa", 2)) # 1
# print(findMaxDistinctCharacters("abcdef", 4)) # 1
```

Problem Statement

Given an integer array numbers and an integer windowSize, determine whether there exists any subarray of size windowSize that contains only distinct elements.

Return:

True if such a window exists.
False otherwise.
Example 1

Input

numbers = [1, 2, 3, 1, 4]
windowSize = 3

Windows

[1, 2, 3] → all distinct ✅

Output

True

```python
def hasDistinctWindow(numbers, windowSize):
    if len(numbers) < windowSize:
        return False
    
    sourceDict = {}
    left = 0
    for right in range(len(numbers)):
        sourceDict[numbers[right]] = sourceDict.get(numbers[right], 0) + 1
        # print(sourceDict, right)
        if right - left + 1 > windowSize:
            sourceDict[numbers[left]] -= 1
            if sourceDict[numbers[left]] == 0:
                del sourceDict[numbers[left]]
            left += 1
        if right - left + 1 == windowSize:
            if len(sourceDict) == windowSize:
                return True

    return False
    
print(hasDistinctWindow([1, 2, 3, 1, 4], 3))   # True
# print(hasDistinctWindow([1, 2, 2, 3], 3))      # False
# print(hasDistinctWindow([5, 5, 5, 5], 2))      # False
# print(hasDistinctWindow([1, 2, 1, 3], 2))      # True
# print(hasDistinctWindow([2, 2, 2, 2, 4], 3))   # False
```

Longest Subarray with Sum Less Than or Equal to Target

Write a function:

findLongestSubarray(array, target)

that returns the length of the longest contiguous subarray whose sum is less than or equal to target.

Example 1
array = [1, 2, 1, 0, 1, 1, 0]
target = 4

Output:
5

Explanation:

The longest valid subarray is

[1, 0, 1, 1, 0]

whose sum is 3 and length is 5.

```python
def findLongestSubarray(array, target):
    left = 0
    currentSum = 0
    maxLength = 0
    
    for right in range(len(array)):
        currentSum += array[right]
        while currentSum > target:
            currentSum = currentSum - array[left]
            left += 1
        maxLength = max(maxLength, right - left + 1)
    
    return maxLength

print(findLongestSubarray([1,2,1,0,1,1,0], 4))
# print(findLongestSubarray([2, 1, 3, 2], 4))
```
