Problem: Count Down

```python
def countDown(n):
    if n == 0:
        return
    print(n)
    countDown(n - 1)
    
countDown(5)
```

Problem: Count Up

```python
def countUp(n):
    if n == 0:
        return
    countUp(n - 1)
    print(n)
    
countUp(5)
```

Problem: Sum of Numbers

```python
def sumNumbers(n):
    if n < 1:
        return 0
    return n + sumNumbers(n - 1)

print(sumNumbers(5))
```

Problem: Factorial

```python
def factorial(n):
    if n <= 1:
        return 1
    return factorial(n - 1) * n

print(factorial(5))
```
