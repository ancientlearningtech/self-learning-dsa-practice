Problem: Count Down

```python
def countDown(n):
    if n == 0:
        return
    print(n)
    countDown(n - 1)
    
countDown(5)
```

Problem: Count Up

```python
def countUp(n):
    if n == 0:
        return
    countUp(n - 1)
    print(n)
    
countUp(5)
```

Problem: Sum of Numbers

```python
def sumNumbers(n):
    if n < 1:
        return 0
    return n + sumNumbers(n - 1)

print(sumNumbers(5))
```

Problem: Factorial

```python
def factorial(n):
    if n <= 1:
        return 1
    return factorial(n - 1) * n

print(factorial(5))
```

Problem: Power of a Number

```python
def power(base, exponent):
    if exponent == 0:
        return 1
    return base * power(base, exponent - 1)

print(power(2, 5))
print(power(2, 0))
```

Problem: Reverse a String

```python
def reverseString(text):
    if len(text) == 0:
        return ''
    return text[-1] + reverseString(text[:-1])
    
print(reverseString('hello'))
```

Problem: Palindrome Check

```python
def isPalindrome(text):
    if len(text) <= 1:
        return True
    if text[0] != text[-1]:
        return False
    return isPalindrome(text[1:-1])

print(isPalindrome("racecar"))
print(isPalindrome("level"))
print(isPalindrome("down"))
print(isPalindrome("abba"))
print(isPalindrome("hello"))
print(isPalindrome(""))
```

Problem: Find Maximum in an Array

```python
def findMaximum(numbers):
    if len(numbers) == 1:
        return numbers[0]
    return max(numbers[0], findMaximum(numbers[1:]))

print(findMaximum([3, 7, 2, 9, 4]))
```

Problem: Sum an Array Recursively

```python
def sumArray(numbers, index):
    if index == 0:
        return numbers[0]
    return numbers[index] + sumArray(numbers, index - 1)

print(sumArray([1,2,3,4,5], 4))
```
