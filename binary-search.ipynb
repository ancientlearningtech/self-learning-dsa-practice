Problem 1: Classic Binary Search

This is the easiest Binary Search problem and forms the foundation for almost every variation.

Problem Statement

You are given a sorted array of integers in ascending order and a target integer.

Return the index of the target if it exists.

If the target does not exist, return -1.

Example 1
Input:
numbers = [2, 5, 8, 11, 16, 20, 27]
target = 16

Output:
4

Example 2
Input:
numbers = [1, 3, 5, 7, 9]
target = 6

Output:
-1

```python
def findTargetIndex(numbers, target):
    
    left = 0
    right = len(numbers) - 1
    mid = left + (right - left) // 2
    
    while left <= right:
        print(left, mid, right)
        if target < numbers[mid]:
            right = mid - 1
            mid = left + (right - left) // 2
        elif target > numbers[mid]:
            left = mid + 1
            mid = left + (right - left) // 2
        else:
            return mid
        
    return -1

# print(findTargetIndex([1,3,5,7,9], 6))
# print(findTargetIndex([2, 5, 8, 11, 16, 20, 27], 16))
print(findTargetIndex([1,3,5,6], 2))
```

Problem: Search Insert Position
Problem Statement

You are given a sorted array of distinct integers and a target value.

If the target exists in the array, return its index.
If the target does not exist, return the index where it should be inserted to keep the array sorted.

You must solve the problem in O(log n) time.

Example 1
Input:
numbers = [1, 3, 5, 6]
target = 5

Output:
2

Explanation:

5 already exists at index 2.

Example 2
Input:
numbers = [1, 3, 5, 6]
target = 2

Output:
1

```python
def searchInsertPosition(numbers, target):
    left = 0
    right = len(numbers) - 1
    mid = left + (right - left) // 2
    
    while left <= right:
        if target < numbers[mid]:
            right = mid - 1
            mid = left + (right - left) // 2
        elif target > numbers[mid]:
            left = mid + 1
            mid = left + (right - left) // 2
        else:
            return mid
        
    return left

print(searchInsertPosition([1,3,5,6], 2))
print(searchInsertPosition([1,3,5,6], 5))
```

Problem: Find First Occurrence of an Element
Problem Statement

You are given a sorted array of integers that may contain duplicate values.

Return the index of the first occurrence of the target.

If the target is not present, return -1.

Your algorithm must run in O(log n) time.

Example 1
Input:
numbers = [1, 2, 2, 2, 3, 4]
target = 2

Output:
1

Explanation:

2 appears at indices:

1, 2, 3

Return the first one.

Example 2
Input:
numbers = [1, 1, 1, 1]
target = 1

Output:
0

```python
def findFirstOccurrence(numbers, target):
    left = 0
    right = len(numbers) - 1
    result = -1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            result = mid
            right = mid - 1
        
    return result

print(findFirstOccurrence([1,2,2,2,3,4], 2))
print(findFirstOccurrence([1,2,3,4,5], 3))
```

```python
def findLastOccurrence(numbers, target):
    left = 0
    right = len(numbers) - 1
    result = -1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            result = mid
            left = mid + 1
        
    return result

print(findLastOccurrence([1,2,2,2,3,4], 2))
print(findLastOccurrence([1,2,3,4,5], 3))
```
