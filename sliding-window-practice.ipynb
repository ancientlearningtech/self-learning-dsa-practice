Problem: Maximum Average Subarray I

Given an integer array nums and an integer k, find the maximum average value of any contiguous subarray of length k.

Example:

Input:
nums = [1,12,-5,-6,50,3]
k = 4

Output:
12.75

Explanation:

Subarray: [12,-5,-6,50]

Sum = 51

Average = 51 / 4 = 12.75

```python
def findMaxAverage(nums, k):
    currentSum = sum(nums[:k])
    maxSum = currentSum

    for right in range(k, len(nums)):
        currentSum = currentSum - nums[right - k] + nums[right]
        maxSum = max(maxSum, currentSum)

    return maxSum / k


print(findMaxAverage([1, 12, -5, -6, 50, 3], 4))
```

Problem #2: Maximum Sum Subarray of Size K
Problem

Given an array of integers nums and an integer k, return the maximum sum of any contiguous subarray of size k.

Example
nums = [2, 1, 5, 1, 3, 2]
k = 3

Output:

9

Explanation:

The subarray:

[5, 1, 3]

has the maximum sum:

5 + 1 + 3 = 9


```python
def maxSubArraySum(nums, k):
    maxSum = sum(nums[:k])
    currentSum = maxSum
    
    for right in range(k, len(nums)):
        currentSum = currentSum - nums[right - k] + nums[right]
        maxSum = max(maxSum, currentSum)
        
    return maxSum
    
print(maxSubArraySum([2, 1, 5, 1, 3, 2], 3))
```

Easy Sliding Window Problem #3: Contains Duplicate II

Given an integer array nums and an integer k, return True if there are two equal numbers whose indices differ by at most k.

Otherwise, return False.

Example 1
nums = [1, 2, 3, 1]
k = 3

Output:

True

Because the two 1s are 3 indices apart.

Example 2
nums = [1, 0, 1, 1]
k = 1

Output:

True

```python
def containsDuplicate(nums, k):
    if len(nums) <= 1:
        return False
    
    window = set()
    left = 0
    
    for right in range(len(nums)):
        if nums[right] in window:
            return True

        window.add(nums[right])
        
        if right - left >= k:
            window.remove(nums[left])
            left += 1
        
    return False
        
print(containsDuplicate([1,2,3,4,1], 3))        
```

Easy Sliding Window Problem #4: Find All Anagrams in a String
Problem

Given two strings:

s
p

Return a list of the starting indices of all anagrams of p in s.

An anagram is a rearrangement of letters.

Example
s = "cbaebabacd"
p = "abc"

Output:

[0, 6]

Why?

Window "cba" (starting at index 0) is an anagram of "abc".

Later, window "bac" (starting at index 6) is also an anagram.

```python
def findAnagrams(s, p):
    if len(p) > len(s):
        return []
    
    output = []
    left = 0
    targetDict = dict()
    sourceDict = dict()
    
    for char in p:
        targetDict[char] = targetDict.get(char, 0) + 1
    
    for chars in s[:len(p)]:
        sourceDict[chars] = sourceDict.get(chars, 0) + 1
        
    if sourceDict == targetDict:
        output.append(0)
        
    for right in range(len(p), len(s)):
        sourceDict[s[right]] = sourceDict.get(s[right], 0) + 1
        sourceDict[s[left]] -= 1
        if sourceDict[s[left]] == 0:
            del sourceDict[s[left]]
        left += 1
        
        if sourceDict == targetDict:
            output.append(left)
            
    return output

print(findAnagrams('cbaebabacd', 'abc')) # [0,6]
print(findAnagrams('abab', 'ab')) # [0,1,2]
```

Permutation in String (Easy)

Problem Statement

Given two strings s1 and s2, return True if s2 contains a permutation (anagram) of s1, otherwise return False.

Example 1
s1 = "ab"
s2 = "eidbaooo"

Output:

True

Because:

"ba"

is a permutation of "ab".

```python
def containsPermutation(s, p):
    if len(s) < len(p):
        return False
    
    left = 0
    sourceDict = dict()
    targetDict = dict()
    
    for char in p:
        targetDict[char] = targetDict.get(char, 0) + 1
    
    for chars in s[:len(p)]:
        sourceDict[chars] = sourceDict.get(chars, 0) + 1
        
    if sourceDict == targetDict:
        return True
        
    for right in range(len(p), len(s)):
        sourceDict[s[right]] = sourceDict.get(s[right], 0) + 1
        sourceDict[s[left]] -= 1
        
        if sourceDict[s[left]] == 0:
            del sourceDict[s[left]]
        left += 1
        
        if sourceDict == targetDict:
            return True

    return False

print(containsPermutation('eabcdbcea', 'ab')) # True
```

Problem #6 — Maximum Number of Vowels in a Substring of Given Length
Problem Statement

Given a string s and an integer k, return the maximum number of vowels in any substring of length k.

Vowels are:

a, e, i, o, u
Example 1
s = "abciiidef"
k = 3

Output:

3

Explanation:

The substring:

"iii"

contains 3 vowels.

```python
def findMaxVowels(s, k):
    if len(s) < k:
        return 0

    left = 0
    vowels = "aeiou"

    currentVowels = 0

    for char in s[:k]:
        if char in vowels:
            currentVowels += 1

    maxVowels = currentVowels

    for right in range(k, len(s)):
        if s[left] in vowels:
            currentVowels -= 1
            
        if s[right] in vowels:
            currentVowels += 1

        left += 1

        maxVowels = max(maxVowels, currentVowels)

    return maxVowels


print(findMaxVowels("abciiidef", 3))   # 3
print(findMaxVowels("aeiou", 2))       # 2
print(findMaxVowels("leetcode", 3))    # 2
```

Problem #7 — Get Equal Substrings Within Budget (Easy–Medium)

You are given two strings s and t of the same length, and an integer maxCost.

Changing s[i] into t[i] costs:

abs(ord(s[i]) - ord(t[i]))

Return the maximum length of a substring that can be transformed from s to t with a total cost less than or equal to maxCost.

Example 1
s = "abcd"
t = "bcdf"
maxCost = 3

Cost at each position:

a → b = 1
b → c = 1
c → d = 1
d → f = 2

Cost array:

[1, 1, 1, 2]

Output:

3

because:

[1, 1, 1] = 3

fits within the budget.

```python
def getEqualSubstring(s, t, maxCost):
    costArray = []

    for i in range(len(s)):
        costArray.append(abs(ord(s[i]) - ord(t[i])))
        
    left = 0
    currentCost = 0
    maxLength = 0

    for right in range(len(costArray)):
        currentCost += costArray[right]
        while currentCost > maxCost:
            currentCost -= costArray[left]
            left += 1
        maxLength = max(maxLength, right - left + 1)

    return maxLength


print(getEqualSubstring("abcd", "bcdf", 3))   # 3
print(getEqualSubstring("abcd", "cdef", 3))   # 1
print(getEqualSubstring("abcd", "acde", 0))   # 1
```

Problem

Given a binary array (0s and 1s) and an integer k, return the length of the longest subarray containing only 1s after flipping at most k zeros.

Example:

nums = [1,1,1,0,0,1,1,0,1]
k = 2

Output: 7

You may flip at most two 0s into 1s.

```python
def longestOnesAfterFlips(nums, k):
    left = 0
    maxLength = 0
    zeroCount = 0
    
    for right in range(len(nums)):
        if nums[right] == 0:
            zeroCount += 1
        
        while zeroCount > k:
            if nums[left] == 0:
                zeroCount -= 1
            left += 1
            
        maxLength = max(maxLength, right - left + 1)
    
    return maxLength
            
print(longestOnesAfterFlips([1,1,1,0,0,1,0,1], 2))
```
