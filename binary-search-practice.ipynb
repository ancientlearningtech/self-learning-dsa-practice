Problem: Given a sorted array of numbers and a target number, return the index of the target.

If the target doesn't exist, return -1.

Example:

numbers = [1, 3, 5, 7, 9, 11]
target = 7

Output: 3

Another:

numbers = [1, 3, 5, 7, 9, 11]
target = 6

Output: -1

```python
def binarySearch(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            return mid
    
    return -1

print(binarySearch([1,3,5,7,9,11], 6))
print(binarySearch([1,3,5,7,9,11], 7))
```

Problem 2: First Occurrence

Given a sorted array that may contain duplicates, find the first occurrence of a target.

Return its index. If the target doesn't exist, return -1.

Examples:

numbers = [1, 2, 2, 2, 4, 5]
target = 2

Output: 1

```python
def findFirstOccurrence(numbers, target):
    left = 0
    right = len(numbers) - 1
    index = -1
    while left <= right:
        mid = left + (right -left) // 2
        # print(left, mid, right)
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            index = mid
            right = mid - 1
        # print(left, mid, right)
    return index

print(findFirstOccurrence([1,2,2,2,2,4,5], 2))
# print(findFirstOccurrence([1,2,2,2,4,5], 2))
# print(findFirstOccurrence([1,2,3,4,5], 6))
# print(findFirstOccurrence([1,2,2,2,4,5], 4))
```

Problem: Last Occurrence

Given a sorted array with duplicates, return the index of the last occurrence of target.

If the target doesn't exist, return -1.

Examples:

numbers = [1, 2, 2, 2, 2, 4, 5]
target = 2

Output: 4

```python
def findLastOccurrence(numbers, target):
    left = 0
    right = len(numbers) - 1
    index = -1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            index = mid
            left = mid + 1
    
    return index

print(findLastOccurrence([1,2,2,2,2,4,5], 2))
```

Problem 4 — Count Occurrences

Now let's combine the two.

Given a sorted array with duplicates, return the number of times target occurs.

Example:

numbers = [1, 2, 2, 2, 2, 4, 5]
target = 2

Output: 4

Another:

numbers = [1, 2, 2, 2, 4, 5]
target = 3

Output: 0

```python
def countOccurrences(numbers, target):
    if findLastOccurrence(numbers, target) == - 1 and findFirstOccurrence(numbers, target) == -1:
        return 0
    return findLastOccurrence(numbers, target) - findFirstOccurrence(numbers, target) + 1

print(countOccurrences([1,2,2,2,4,5], 2))
print(countOccurrences([1,2,2,2,4,5], 6))
```

Problem: Search Insert Position

Given a sorted array of distinct numbers and a target, return:

the index if the target exists
otherwise, the index where the target should be inserted so the array remains sorted.
Examples
numbers = [1, 3, 5, 6]
target = 5

Output: 2
numbers = [1, 3, 5, 6]
target = 2

Output: 1

```python
def searchInsertPosition(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            return mid
    
    return left

print(searchInsertPosition([1,3,5,6], 2)) # 1
# print(searchInsertPosition([1,3,5,6], 3)) # 1
# print(searchInsertPosition([1,3,5,6], 6)) # 3
# print(searchInsertPosition([1,3,5,6], 7)) # 4
# print(searchInsertPosition([], 7)) # 0
```
