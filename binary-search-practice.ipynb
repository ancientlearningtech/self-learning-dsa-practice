Problem: Given a sorted array of numbers and a target number, return the index of the target.

If the target doesn't exist, return -1.

Example:

numbers = [1, 3, 5, 7, 9, 11]
target = 7

Output: 3

Another:

numbers = [1, 3, 5, 7, 9, 11]
target = 6

Output: -1

```python
def binarySearch(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            return mid
    
    return -1

print(binarySearch([1,3,5,7,9,11], 6))
print(binarySearch([1,3,5,7,9,11], 7))
```

Problem 2: First Occurrence

Given a sorted array that may contain duplicates, find the first occurrence of a target.

Return its index. If the target doesn't exist, return -1.

Examples:

numbers = [1, 2, 2, 2, 4, 5]
target = 2

Output: 1

```python
def findFirstOccurrence(numbers, target):
    left = 0
    right = len(numbers) - 1
    index = -1
    while left <= right:
        mid = left + (right -left) // 2
        # print(left, mid, right)
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            index = mid
            right = mid - 1
        # print(left, mid, right)
    return index

print(findFirstOccurrence([1,2,2,2,2,4,5], 2))
# print(findFirstOccurrence([1,2,2,2,4,5], 2))
# print(findFirstOccurrence([1,2,3,4,5], 6))
# print(findFirstOccurrence([1,2,2,2,4,5], 4))
```

Problem: Last Occurrence

Given a sorted array with duplicates, return the index of the last occurrence of target.

If the target doesn't exist, return -1.

Examples:

numbers = [1, 2, 2, 2, 2, 4, 5]
target = 2

Output: 4

```python
def findLastOccurrence(numbers, target):
    left = 0
    right = len(numbers) - 1
    index = -1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            index = mid
            left = mid + 1
    
    return index

print(findLastOccurrence([1,2,2,2,2,4,5], 2))
```

Problem 4 — Count Occurrences

Now let's combine the two.

Given a sorted array with duplicates, return the number of times target occurs.

Example:

numbers = [1, 2, 2, 2, 2, 4, 5]
target = 2

Output: 4

Another:

numbers = [1, 2, 2, 2, 4, 5]
target = 3

Output: 0

```python
def countOccurrences(numbers, target):
    if findLastOccurrence(numbers, target) == - 1 and findFirstOccurrence(numbers, target) == -1:
        return 0
    return findLastOccurrence(numbers, target) - findFirstOccurrence(numbers, target) + 1

print(countOccurrences([1,2,2,2,4,5], 2))
print(countOccurrences([1,2,2,2,4,5], 6))
```

Problem: Search Insert Position

Given a sorted array of distinct numbers and a target, return:

the index if the target exists
otherwise, the index where the target should be inserted so the array remains sorted.
Examples
numbers = [1, 3, 5, 6]
target = 5

Output: 2
numbers = [1, 3, 5, 6]
target = 2

```python
def searchInsertPosition(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if target < numbers[mid]:
            right = mid - 1
        elif target > numbers[mid]:
            left = mid + 1
        else:
            return mid
    
    return left

print(searchInsertPosition([1,3,5,6], 2)) # 1
# print(searchInsertPosition([1,3,5,6], 3)) # 1
# print(searchInsertPosition([1,3,5,6], 6)) # 3
# print(searchInsertPosition([1,3,5,6], 7)) # 4
# print(searchInsertPosition([], 7)) # 0
```

Problem: Find the Floor

Given a sorted array of distinct numbers and a target, find the largest number that is less than or equal to the target.

If no such number exists, return -1.

Examples
numbers = [1, 3, 5, 7, 9]
target = 6

Output: 5

Because 5 is the largest number ≤ 6.

numbers = [1, 3, 5, 7, 9]
target = 7

Output: 7

```python
def findFloor(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if numbers[mid] > target:
            right = mid - 1
        elif numbers[mid] < target:
            left = mid + 1
        else:
            return numbers[mid]
    
    if right == -1:
        return -1
    else:
        return numbers[right] 
    
print(findFloor([1,3,5,7,9], 7))
# print(findFloor([1,3,5,7,9], 6))
# print(findFloor([1,3,5,7,9], 0))
# print(findFloor([1,3,5,7,9], 2))
```

Problem: Find the Ceiling

Let's do the exact opposite now.

Given a sorted array of distinct numbers, find the smallest number that is greater than or equal to target.

Examples:

numbers = [1, 3, 5, 7, 9]
target = 6

Output: 7
numbers = [1, 3, 5, 7, 9]
target = 7

Output: 7
numbers = [1, 3, 5, 7, 9]
target = 8

Output: 9
numbers = [1, 3, 5, 7, 9]
target = 10

Output: -1

```python
def findCeiling(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        if numbers[mid] > target:
            right = mid - 1
        elif numbers[mid] < target:
            left = mid + 1
        else:
            return numbers[mid]
    # print(left, mid, right)
    if left == len(numbers):
        return -1
    else:
        return numbers[left]
    
print(findCeiling([1,3,5,7,9], 6)) # 7
# print(findCeiling([1,3,5,7,9], 7)) # 7
# print(findCeiling([1,3,5,7,9], 8)) # 9
# print(findCeiling([1,3,5,7,9], 10)) # -1
```

Problem: Find the closest number

Now let's make it slightly more interesting.

Given a sorted array of distinct numbers and a target, return the number that is closest to the target.

Examples:

numbers = [1, 3, 5, 7, 9]
target = 6

Output: 5

5 and 7 are equally close to 6.

For this problem, if there is a tie, return the smaller number.

Another:

numbers = [1, 3, 5, 7, 9]
target = 8

Output: 7

And:

numbers = [1, 3, 5, 7, 9]
target = 10

Output: 9

```python
def findClosestNumber(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <=right:
        mid = left + (right - left) // 2
        
        if numbers[mid] < target:
            left = mid + 1
        elif numbers[mid] > target:
            right = mid - 1
        else:
            return numbers[mid]
    
    # print(left, mid, right)
    # if left == len(numbers):
    #     return numbers[left - 1]
    # elif numbers[left] > numbers[right]:
    #     return numbers[right]
    # else:
    #     return numbers[left]
    
    if right == -1:
        return numbers[0]
    elif left == len(numbers):
        return numbers[-1]
    elif abs(numbers[left] - target) < abs(numbers[right] - target):
        return numbers[left]
    else:
        return numbers[right]
    
print(findClosestNumber([1,3,5,7,9], 10))
print(findClosestNumber([1,3,5,7,9], 6))
print(findClosestNumber([1,3,5,7,9], 7))
print(findClosestNumber([1,3,5,7,9], 0))
print(findClosestNumber([1,3,5,7,9], 22))
```

Problem: Find the Peak Element

Given an array of numbers, find an element that is greater than its immediate neighbours.

Return the index of the peak element.

For the purposes of this problem, assume:

numbers[-1] = -∞
numbers[len(numbers)] = -∞

So the first or last element can also be a peak.

Examples
numbers = [1, 2, 3, 1]

Output: 2

Because:

1 < 3 > 1

Another:

numbers = [1, 2, 1, 3, 5, 6, 4]

Output: 5

Because:

5 < 6 > 4

```python
def findPeakElement(numbers):
    left = 0
    right = len(numbers) - 1
    
    while left < right:
        mid = left + (right - left) // 2
        
        if numbers[mid] < numbers[mid + 1]:
            left = mid + 1
        else:
            return mid
    
print(findPeakElement([1,2,1,3,5,6,4]))
print(findPeakElement([1,2,3,1]))
```

Problem: Search in a Rotated Sorted Array

This is a classic interview Binary Search problem.

You are given a sorted array that has been rotated at some point.

For example:

Original:
[1, 2, 3, 4, 5, 6, 7]

Rotated:
[4, 5, 6, 7, 1, 2, 3]

The array contains distinct numbers.

Given a target, return its index. If it doesn't exist, return -1.

Examples
numbers = [4, 5, 6, 7, 0, 1, 2]
target = 0

Output: 4
numbers = [4, 5, 6, 7, 0, 1, 2]
target = 3

Output: -1
numbers = [6, 7, 1, 2, 3, 4, 5]
target = 4

Output: 5

```python
def searchRotatedArray(numbers, target):
    left = 0
    right = len(numbers) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        
        if numbers[mid] == target:
            return mid
        if numbers[left] > numbers[mid]:  # search right half
            if numbers[mid] < target <= numbers[right]:
                left = mid + 1
            else:
                right = mid - 1
        else:                            # search left half
            if numbers[left] <= target < numbers[mid]:
                right = mid - 1
            else:
                left = mid + 1
        
    return -1

print(searchRotatedArray([4,5,6,7,0,1,2], 0))  # 4
# print(searchRotatedArray([4,5,6,7,0,1,2], 3))  # -1
# print(searchRotatedArray([6,7,1,2,3,4,5], 4))  # 5
# print(searchRotatedArray([1], 1))              # 0
# print(searchRotatedArray([1], 2))              # -1
```
