```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
```

```python
from collections import deque
def levelOrderTraversal(root):
    queue = deque([root])
    result = []
    
    while queue:
        levelSize = len(queue)
        currentLevel = []
        
        for _ in range(levelSize):
            node = queue.popleft()
            currentLevel.append(node.value)
            
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        result.append(currentLevel)
        
    return result

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.right.left = TreeNode(6)

print(levelOrderTraversal(root))
```

Problem: Minimum Depth of a Binary Tree

Given a binary tree, find the minimum depth of the tree.

What is depth?

The depth is the number of nodes along the path from the root to a leaf.

A leaf is a node that has:

left  = None
right = None

For example:

        1
       / \
      2   3
     /
    4

There are two root-to-leaf paths:

1 → 2 → 4     depth = 3
1 → 3         depth = 2

Therefore:

minimum depth = 2

```python
from collections import deque
def findMinimumDepth(root):
    if root is None:
        return 0
    queue = deque([root])
    depth = 0
    
    while queue:
        levelSize = len(queue)
        depth += 1
        
        for _ in range(levelSize):
            node = queue.popleft()
            
            if node.left is None and node.right is None:
                return depth
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)

print(findMinimumDepth(root))
```
